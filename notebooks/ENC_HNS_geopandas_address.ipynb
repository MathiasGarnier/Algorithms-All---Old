*Développement d'une chaine de traitement pour l'analyse de l'Annuaire des propriétaires et des propriétés de Paris et du département de la Seine (1898, 1903, 1913, 1923, 1951)*

Les annuaires et les répertoires anciens constituent une source riche d’information sur un moment précis de la géographie des villes historiques, mettant en rapport la distribution de l’espace avec des données socio-économiques ou professionnelles. ###Cartographie thématique

**Carte cercles proportionnels - stock**

Définition type de variable stock : il s'agit d'une valeur quantitative absolue comme la nombre d'habitants, l'age, ou la production totale d'une marchandise (Magrit).

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 1, figsize=(12, 10))

# Fond : quartiers
addr_count_quartier.plot(ax=ax, color='lightgray', edgecolor='black', linewidth=0.5)

# Centroïdes
centroids = addr_count_quartier.geometry.centroid

# Cercles proportionnels
FACTEUR = 20  # avec 2, les cercles seraient minuscules (7 → 14 pt², 100 → 200 pt²)

scatter = ax.scatter(
    centroids.x,
    centroids.y,
    s=addr_count_quartier['count_addr'] * FACTEUR,
    color='tomato',
    alpha=0.6,
    edgecolors='black',
    linewidth=0.5
)

# Légende adaptée à vos données
valeurs_legende = [10, 50, 100]
handles = [
    ax.scatter([], [], s=v * FACTEUR, color='tomato', alpha=0.6,
               edgecolors='black', linewidth=0.5)
    for v in valeurs_legende
]
ax.legend(
    handles, [str(v) for v in valeurs_legende],
    title="Nombre d'adresses",
    loc='lower right',
    labelspacing=2,
    borderpad=1.2,
    scatterpoints=1
)

plt.title("Nombre d'adresses par quartier à Paris (1898)", fontsize=16, pad=20)
plt.axis('off')

plt.tight_layout()
plt.show()
```

**Carte choroplèthe - rapport ou ratio**

Définition type de variable rapport (ou ratio): Les données de rapports (de taux, d’intensités ou de ratios) sont des données quantitatives calculées à partir de données de stocks dont ont fait le rapport ou dont on calcule le taux à partir d’un total. **Elles expriment les caractéristiques des individus observés mais leur total n'a pas de signification concrète**. (Magrit)

```python
import matplotlib.pyplot as plt

# Calculer la superficie en km²
addr_count_quartier['superficie_km2'] = addr_count_quartier.geometry.area / 1_000_000

# Calculer la densité d'adresses par km²
addr_count_quartier['densite_km2'] = addr_count_quartier['count_addr'] / addr_count_quartier['superficie_km2']

# Carte choroplèthe
fig, ax = plt.subplots(1, 1, figsize=(12, 10))

addr_count_quartier.plot(
    ax=ax,
    column='densite_km2',
    cmap='YlOrRd',
    legend=True,
    edgecolor='black',
    linewidth=0.5,
    legend_kwds={'shrink': 0.8}
)

# Étiquettes : valeur de c_ar au centre de chaque polygone
for idx, row in addr_count_quartier.iterrows():
    point = row.geometry.representative_point()  # toujours à l'intérieur du polygone
    ax.annotate(
        text=int(row['c_ar']),
        xy=(point.x, point.y),
        ha='center',
        va='center',
        fontsize=8,
        color='black'
    )

# Ajouter manuellement le label de la légende
cbar = ax.get_figure().axes[-1]  # Récupérer l'axe de la colorbar
cbar.set_ylabel('Densité (adresses/km²)', rotation=270, labelpad=20)

plt.title('Densité d\'adresses par quartier à Paris (1898)', fontsize=16, pad=20)
plt.axis('off')
plt.tight_layout()
plt.show()
```

**EQUAL INTERVAL** divise les données en classes de taille égale (par exemple, 0-10, 10-20, 20-30, etc.) et fonctionne mieux avec des données qui sont généralement réparties sur toute la plage.

ATTENTION : évitez l'intervalle égal si vos données tendent à une extrémité ou si vous avez une ou deux valeurs aberrantes vraiment importantes.

```python
import mapclassify

mapclassify.EqualInterval(addr_count_quartier.count_addr, k=5)
```

**QUANTILES** essaie de classer le même nombre d'observations dans chacune des N classes. En d'autres termes, cette méthode essaie d'organiser les groupes de façon à ce qu'ils aient la même quantité. Le problème avec les quantiles est que vous pouvez vous retrouver avec des classes qui ont des plages numériques très différentes (par exemple, 1-4, 4-9, 9-250).

```python
mapclassify.Quantiles(addr_count_quartier.count_addr, k=5)
```

**NATURAL BREAKS (Jenks)** organise chaque groupe de manière à ce qu'il y ait moins de variation dans chaque classe ou nuance.

est une sorte de schéma de classification "optimal" qui trouve les ruptures de classe qui minimiseront la variance intra-classe et maximiseront les différences entre les classes. Un inconvénient de cette approche est que chaque ensemble de données génère une solution de classification unique, et si vous avez besoin de faire des comparaisons entre les cartes, comme dans un atlas ou une série (par exemple, une carte pour 1980, 1990, 2000), vous voudrez peut-être utiliser un seul schéma qui peut être appliqué à toutes les cartes.



```python
mapclassify.NaturalBreaks(addr_count_quartier.count_addr)
```

### Où habitent les propriétaires ?

Utilisez le service pour géocoder des adresses françaises : https://adresse.data.gouv.fr/outils/csv pour géocoder les adresses d'habitation des propriétaires (colonnes concernées : num_dom, rue_dom, voie_dom, ville_dom) puis uploadez le fichier résultant sur le panel à gauche, le système de fichiers.

```python
import pandas as pd
import geopandas as gpd
import numpy as np
import matplotlib.pyplot as plt
from shapely.geometry import LineString

# --- Fichiers d'entrée ---
FICHIER_PROPRIETE = "selection_addr_prop_Paris.geocoded.csv"    # adresse de l'immeuble géocodée
FICHIER_HABITATION = "selection_addr_prop_Paris_geocoded_addrhabitation.csv" # adresse du propriétaire géocodée
SEP = ","

# --- Colonnes ---
COL_ID = "n_order"          # clé commune aux deux fichiers
COL_LON, COL_LAT = "longitude", "latitude"
COL_SCORE = "result_score"
COL_CITYCODE = "result_citycode"
COL_PERS = "nom_prop"

# --- Filtres de qualité du géocodage ---
SCORE_MIN = 0.5                                 # None pour désactiver
BBOX_PARIS = (2.22, 48.81, 2.47, 48.91)         # lon_min, lat_min, lon_max, lat_max
GARDER_HORS_PARIS_HABITATION = False            # True : garder les propriétaires hors Paris

# --- Analyse ---
CRS_METRIQUE = 2154       # Lambert 93
SEUIL_SUR_PLACE = 100     # m : en dessous, le propriétaire "habite sur place"
COL_ARR = "c_ar"          # colonne de l'arrondissement dans addr_count_quartier
```

Chargement des données et sélection des paires de coordonnées pour préparer le calcul des distances.

```python
def charger(chemin, suffixe, filtrer_bbox=True):
    df = pd.read_csv(chemin, sep=SEP, low_memory=False)
    n0 = len(df)
    df = df.dropna(subset=[COL_LON, COL_LAT])
    if SCORE_MIN is not None and COL_SCORE in df:
        df = df[df[COL_SCORE] >= SCORE_MIN]
    if filtrer_bbox:
        x0, y0, x1, y1 = BBOX_PARIS
        df = df[df[COL_LON].between(x0, x1) & df[COL_LAT].between(y0, y1)]
    print(f"{suffixe}: {len(df)}/{n0} lignes conservées ({len(df)/n0:.0%})")
    cols = [COL_ID, COL_LON, COL_LAT]
    return df[cols].rename(columns={COL_LON: f"lon_{suffixe}", COL_LAT: f"lat_{suffixe}"}), df

prop, prop_full = charger(FICHIER_PROPRIETE, "imm", filtrer_bbox=True)
hab, _ = charger(FICHIER_HABITATION, "hab", filtrer_bbox=not GARDER_HORS_PARIS_HABITATION)

# Informations descriptives de l'immeuble (propriétaire, voie...)
infos = prop_full[[COL_ID, COL_PERS, "nom_voie", "type_voie"]].drop_duplicates(COL_ID)

paires = (prop.merge(hab, on=COL_ID, how="inner")
              .merge(infos, on=COL_ID, how="left"))
print(f"\nPaires propriété/habitation exploitables : {len(paires)}")
```

La fonction charger nettoie un fichier : elle retire les adresses non géocodées, celles dont le score est trop bas et celles hors de Paris, puis renvoie un tableau réduit (identifiant + coordonnées renommées avec un suffixe _imm ou _hab). Le merge colle ensuite les deux fichiers côte à côte grâce à l’identifiant commun nOrder, pour avoir sur une même ligne la position de l’immeuble et celle du propriétaire. inner ne garde que les propriétés présentes dans les deux fichiers.

```python
gdf_imm = gpd.GeoDataFrame(
    paires, geometry=gpd.points_from_xy(paires.lon_imm, paires.lat_imm), crs=4326
).to_crs(CRS_METRIQUE)

gdf_hab = gpd.GeoDataFrame(
    paires, geometry=gpd.points_from_xy(paires.lon_hab, paires.lat_hab), crs=4326
).to_crs(CRS_METRIQUE)

paires["dist_m"] = gdf_imm.geometry.distance(gdf_hab.geometry).values
paires["sur_place"] = paires["dist_m"] <= SEUIL_SUR_PLACE

print(paires["dist_m"].describe().round(0))
print(f"\nPropriétaires habitant sur place (<= {SEUIL_SUR_PLACE} m) : {paires['sur_place'].mean():.1%}")
```

On transforme les colonnes lon/lat en vrais points géographiques (GeoDataFrame), puis on les convertit en Lambert 93 (to_crs(2154)), car il faut un système en mètres pour mesurer une distance (en degrés, le résultat n’aurait pas de sens). .distance() calcule ensuite, ligne par ligne, l’écart entre chaque immeuble et l’habitation correspondante.

```python
quartiers = addr_count_quartier[[COL_ARR, "geometry"]].to_crs(CRS_METRIQUE)

def arr_de(points, nom):
    j = gpd.sjoin(gpd.GeoDataFrame(geometry=points.geometry.values, crs=CRS_METRIQUE),
                  quartiers, how="left", predicate="within")
    j = j[~j.index.duplicated()]
    return j[COL_ARR].values

paires["arr_imm"] = arr_de(gdf_imm, "imm")
paires["arr_hab"] = arr_de(gdf_hab, "hab")
paires["arr_hab"] = paires["arr_hab"].fillna(-1)   # -1 = hors Paris / hors polygones
paires = paires.dropna(subset=["arr_imm"])
```

sjoin (jointure spatiale) demande, pour chaque point, dans quel polygone d’arrondissement il tombe (within). On retire les doublons d’index au cas où un point toucherait deux polygones, et on met -1 aux propriétaires dont l’habitation n’est dans aucun arrondissement (hors Paris).

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(paires["dist_m"] / 1000, bins=40, color="tomato", edgecolor="black", alpha=0.7)
axes[0].axvline(paires["dist_m"].median() / 1000, color="black", ls="--",
                label=f"médiane = {paires['dist_m'].median()/1000:.2f} km")
axes[0].set_xlabel("Distance propriétaire - immeuble (km)")
axes[0].set_ylabel("Nombre de propriétés")
axes[0].legend()

med = paires.groupby("arr_imm")["dist_m"].median() / 1000
axes[1].bar(med.index.astype(int).astype(str), med.values, color="tomato", edgecolor="black")
axes[1].set_xlabel("Arrondissement de la propriété")
axes[1].set_ylabel("Distance médiane (km)")

plt.tight_layout()
plt.show()
```

À gauche, la répartition des distances en km avec une ligne verticale pour la médiane. À droite, groupby("arr_imm") calcule la distance médiane pour chaque arrondissement, pour voir où les propriétaires vivent le plus loin de leurs biens.

```python
matrice = pd.crosstab(paires["arr_hab"], paires["arr_imm"])
matrice.index = ["Hors Paris" if i == -1 else int(i) for i in matrice.index]
matrice.columns = [int(c) for c in matrice.columns]

# En % : pour chaque arrondissement d'habitation, où se trouvent les biens ?
matrice_pct = matrice.div(matrice.sum(axis=1), axis=0) * 100

fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(matrice_pct.values, cmap="YlOrRd")
ax.set_xticks(range(len(matrice_pct.columns)), matrice_pct.columns)
ax.set_yticks(range(len(matrice_pct.index)), matrice_pct.index)
ax.set_xlabel("Arrondissement de la propriété")
ax.set_ylabel("Arrondissement d'habitation du propriétaire")
plt.colorbar(im, label="% des biens (par ligne)", shrink=0.8)
ax.set_title("Où les propriétaires possèdent-ils selon leur lieu de résidence ?")
plt.tight_layout()
plt.show()
```

pd.crosstab compte combien de propriétés relient chaque arrondissement d’habitation (lignes) à chaque arrondissement de propriété (colonnes). La division par la somme de chaque ligne transforme ces comptes en pourcentages : chaque ligne se lit comme « où possèdent les gens qui habitent ici ? » (elle fait 100 %). imshow affiche ensuite ce tableau sous forme de carte de chaleur.

```python
vals = matrice_pct.drop(index="Hors Paris", errors="ignore").copy()
arr = vals.index.intersection(vals.columns)
data = vals.values.astype(float)
for k, a in enumerate(vals.index):
    if a in vals.columns:
        data[k, list(vals.columns).index(a)] = np.nan   # diagonale masquée

fig, ax = plt.subplots(figsize=(10, 8))
im = ax.imshow(np.ma.masked_invalid(data), cmap="YlOrRd")
ax.set_xticks(range(len(vals.columns)), vals.columns)
ax.set_yticks(range(len(vals.index)), vals.index)
ax.set_xlabel("Arrondissement de la propriété")
ax.set_ylabel("Arrondissement d'habitation")
plt.colorbar(im, label="% des biens (par ligne), hors diagonale", shrink=0.8)
plt.tight_layout()
plt.show()
```

La diagonale correspond aux propriétaires qui possèdent dans leur propre arrondissement et domine souvent l’échelle de couleurs. La boucle la remplace par NaN et masked_invalid dit à matplotlib de ne pas la colorier, pour que les flux entre arrondissements différents ressortent.
